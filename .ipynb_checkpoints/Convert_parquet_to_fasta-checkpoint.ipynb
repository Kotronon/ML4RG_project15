```python
from pathlib import Path
import pyarrow.parquet as pq

project_dir = Path("/s/project/ml4rg_students/2026/project15")
input_dir = project_dir / "raw" / "sequence_datasets"
output_dir = project_dir / "working" /"sequence_datasets_fastas"
output_dir.mkdir(parents=True, exist_ok=True)

parquet_files = sorted(input_dir.glob("*.parquet"))
print("Number of parquet files:", len(parquet_files))
print("Output dir:", output_dir)
```

```python
for parquet_file in parquet_files:
    
    print("Reading:", parquet_file.name)

    table = pq.read_table(
        parquet_file,
        columns=["chrom", "start", "end", "strand", "seq", "gene_id"]
    )
    df = table.to_pandas()

    out_path = output_dir / (parquet_file.stem + ".fasta")
    if out_path.exists():
        continue

    with open(out_path, "w") as out:
        for i, row in df.iterrows():
            seq = str(row["seq"]).upper()
            seq = "".join(c if c in "ACGTN" else "N" for c in seq)

            if len(seq) == 0:
                continue

            header = (
                f"{row['gene_id']}"
                f"|species_file={parquet_file.stem}"
                f"|chr={row['chrom']}"
                f"|start={row['start']}"
                f"|end={row['end']}"
                f"|strand={row['strand']}"
            )

            out.write(f">{header}\n")
            for j in range(0, len(seq), 80):
                out.write(seq[j:j+80] + "\n")

    print("Wrote:", out_path)
```

```python
fasta_files = sorted(output_dir.glob("*.fasta"))
print(len(fasta_files))
```
